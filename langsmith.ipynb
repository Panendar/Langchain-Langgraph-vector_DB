#### Core concepts: Project ==> the whole execution which loops user-> prompt -> LLM -> Parser      Trace ==> the single execution is know as trace     Run ==> the every component is know as run i.e; prompt,llm,parser -> 3 runs


```python
# A simple LLM call
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv()

# simple one line prompt
prompt = PromptTemplate.from_template("{question}")

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.7,
)

parser = StrOutputParser()

# chain: prompt -> model -> parser
chain = prompt | model | parser

# Run it
result = chain.invoke({"question": "What is the capital of peru"})
print(result)
```

```python
# sequential flow

prompt1 = PromptTemplate(
    template='Generate a detailed report on {topic}',
    input_variables=['topic']
)

prompt2 = PromptTemplate(
    template='Generate a 5 pointer summary from the following text \n {text}',
    input_variables=['text']
)

parser = StrOutputParser()

chain_1 = prompt1 | model | parser | prompt2 | model | parser

result_1 = chain_1.invoke({"topic": "Langchain"})
print(result_1)
```
